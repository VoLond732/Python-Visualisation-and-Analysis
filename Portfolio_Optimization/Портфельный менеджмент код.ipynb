```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import scipy.optimize as sco
import os
from datetime import datetime, timedelta

DATA_FILE = r"stocks_python.xlsx"
OUTPUT_FOLDER = r"Portfolio_Optimization"
SELECTED_INDICES_NAMES = ['SBER', 'FLOT', 'IRAO', 'NVTK', 'PHOR', 'MGNT', 'CHMF']
START_DATE_STR = "2021-01-04"
FORECAST_START_DATE_STR = "2025-12-01"
END_DATE_STR = None
FORECAST_YEARS = 20
MIN_ALLOCATION = 0.05
MAX_ALLOCATION = 0.25
NUM_SIMULATIONS = 100000
INITIAL_DEPOSIT = 144637500
ANNUAL_CONTRIBUTION = 45080000
DIVIDEND_YIELDS = {
    'SBER': 0.0801,
    'FLOT': 0.0587,
    'IRAO': 0.0741,
    'NVTK': 0.0646,
    'PHOR': 0.1141,
    'MGNT': 0.0484,
    'CHMF': 0.0734,
}
risk_free_rate = 0.1361

RISK_AVERSION_DECAY = 0.9 
VOL_ROLLING_YEARS = 3
DEBUG_REBALANCE = False

plt.rcParams["figure.figsize"] = (12, 6)
TRADING_DAYS = 252

def load_data(file_path, sheet_name="DATA"):
    try:
        df = pd.read_excel(file_path, sheet_name=sheet_name, index_col=0, parse_dates=True)
        return df
    except FileNotFoundError:
        print(f"Ошибка: Файл '{file_path}' не найден.")
        return None
    except Exception as e:
        print(f"Ошибка при загрузке данных: {e}")
        return None

def select_indices(df, selected_indices_names):
    available_indices = df.columns.tolist()
    selected_indices = [index for index in selected_indices_names if index in available_indices]
    if not selected_indices:
        print(f"В файле не найден ни один из указанных индексов: {selected_indices_names}. Доступные индексы: {available_indices}")
        return None
    return df[selected_indices]

def select_date_range(df, start_date_str=None, end_date_str=None):
    start_date = datetime.strptime(start_date_str, '%Y-%m-%d').date() if start_date_str else df.index.min().date()
    end_date = datetime.strptime(end_date_str, '%Y-%m-%d').date() if end_date_str else df.index.max().date()
    if start_date <= end_date:
        return df[(df.index.date >= start_date) & (df.index.date <= end_date)]
    else:
        print("Начальная дата не может быть позже конечной.")
        return None

def calculate_returns(df):
    price_returns = df.pct_change()
    dividend_daily = {}
    for ticker in df.columns:
        if ticker in DIVIDEND_YIELDS:
            annual_yield = DIVIDEND_YIELDS[ticker]
            daily_yield = (1 + annual_yield) ** (1/252) - 1
            dividend_daily[ticker] = daily_yield
        else:
            dividend_daily[ticker] = 0.0
    div_vec = np.array([dividend_daily[t] for t in df.columns], dtype=float)
    dividend_df = pd.DataFrame(np.tile(div_vec, (len(price_returns), 1)),
                               index=price_returns.index, columns=price_returns.columns)
    total_returns = price_returns + dividend_df
    return total_returns.dropna()

def portfolio_annualised_performance(weights, mean_returns, cov_matrix):
    returns = np.sum(mean_returns * weights) * TRADING_DAYS
    std_dev = np.sqrt(np.dot(weights.T, np.dot(cov_matrix, weights))) * np.sqrt(TRADING_DAYS)
    return std_dev, returns

def neg_sharpe_ratio(weights, mean_returns, cov_matrix, risk_free_rate):
    p_var, p_ret = portfolio_annualised_performance(weights, mean_returns, cov_matrix)
    return -(p_ret - risk_free_rate) / p_var

def markowitz_optimization(returns, risk_free_rate, bounds=(0, 1)):
    n = len(returns.columns)
    if MIN_ALLOCATION * n > 1.0 + 1e-12:
        raise ValueError("MIN_ALLOCATION слишком велик для числа активов.")
    mean_returns = returns.mean()
    cov_matrix = returns.cov()
    num_assets = n
    args = (mean_returns, cov_matrix)
    constraints = ({'type': 'eq', 'fun': lambda x: np.sum(x) - 1})
    bounds = tuple(bounds for asset in range(num_assets))

    def minimize_volatility(weights, mean_returns, cov_matrix):
        return portfolio_annualised_performance(weights, mean_returns, cov_matrix)[0]

    opt_vol = sco.minimize(minimize_volatility, np.array([1/num_assets]*num_assets), args=args, method='SLSQP', bounds=bounds, constraints=constraints)

    def minimize_neg_sharpe(weights, mean_returns, cov_matrix, risk_free_rate):
        return neg_sharpe_ratio(weights, mean_returns, cov_matrix, risk_free_rate)

    opt_sharpe = sco.minimize(minimize_neg_sharpe, np.array([1/num_assets]*num_assets), args=(mean_returns, cov_matrix, risk_free_rate), method='SLSQP', bounds=bounds, constraints=constraints)

    return opt_vol, opt_sharpe

def rand_weights_with_min_max(n, min_w, max_w, rng=np.random, max_tries=5000):
    if min_w * n > 1.0 + 1e-12:
        raise ValueError("min_w * n > 1 — невозможные ограничения.")
    for _ in range(max_tries):
        base = np.full(n, min_w)
        remaining = 1.0 - base.sum()
        if remaining <= 0:
            w = base.copy()
        else:
            y = rng.gamma(1.0, 1.0, size=n)
            y = y / y.sum()
            w = base + remaining * y
        if np.all(w <= max_w + 1e-12):
            return w
    raise RuntimeError("Не удалось сгенерировать веса в заданных границах.")

def monte_carlo_simulation(returns, risk_free_rate, num_simulations=100000, bounds=(0, 1)):
    mean_returns = returns.mean()
    cov_matrix = returns.cov()
    num_assets = len(returns.columns)
    if MIN_ALLOCATION * num_assets > 1.0 + 1e-12:
        raise ValueError("MIN_ALLOCATION слишком велик для числа активов.")

    results = np.zeros((3 + num_assets, num_simulations))
    for i in range(num_simulations):
        weights = rand_weights_with_min_max(num_assets, bounds[0], bounds[1])
        p_vol, p_ret = portfolio_annualised_performance(weights, mean_returns, cov_matrix)
        results[0, i] = p_vol
        results[1, i] = p_ret
        results[2, i] = (p_ret - risk_free_rate) / p_vol
        results[3:, i] = weights
    return results

def calculate_max_drawdown(portfolio_values):
    arr = np.asarray(portfolio_values, dtype=float)
    cummax = np.maximum.accumulate(arr)
    drawdowns = (cummax - arr) / cummax
    return np.nanmax(drawdowns)

def plot_portfolio_forecast(portfolio_values, dates, title, output_path):
    plt.figure(figsize=(12, 6))
    plt.plot(dates, portfolio_values)
    plt.gca().xaxis.set_major_locator(mdates.AutoDateLocator())
    plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
    plt.xlabel('Дата')
    plt.ylabel('Стоимость портфеля, руб.')
    plt.title(title)
    plt.grid(True)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.savefig(output_path)
    plt.show()

def plot_efficient_frontier_and_portfolios(monte_carlo_results, mean_weights, min_vol_port, max_sharpe_port, returns, output_path):
    plt.figure(figsize=(12, 8))

    plt.scatter(monte_carlo_results[0, :] * 100, monte_carlo_results[1, :] * 100, marker='o')
    plt.xlabel('Годовая волатильность, %')
    plt.ylabel('Годовая доходность, %')
    plt.title('Облако портфелей (Монте-Карло)')

    mean_vol, mean_ret = portfolio_annualised_performance(mean_weights['weights'], returns.mean(), returns.cov())
    plt.scatter(mean_vol * 100, mean_ret * 100, marker='X', color='r', s=200, label='Средние доли')

    try:
        min_vol_val = min_vol_port['fun']
    except Exception:
        min_vol_val = None

    if min_vol_val is not None:
        min_ret = portfolio_annualised_performance(min_vol_port['x'], returns.mean(), returns.cov())[1]
        plt.scatter(min_vol_val * 100, min_ret * 100, marker='*', color='g', s=200, label='Мин. риск')

    max_sharpe_vol, max_sharpe_ret = portfolio_annualised_performance(max_sharpe_port['x'], returns.mean(), returns.cov())
    plt.scatter(max_sharpe_vol * 100, max_sharpe_ret * 100, marker='P', color='gold', s=200, label='Макс. Sharpe')

    plt.legend()
    plt.grid(True)
    plt.savefig(output_path)
    plt.show()


def plot_portfolio_structure(weights_history, dates, assets, output_path):
    plt.figure(figsize=(12, 6))
    plt.stackplot(dates, np.array(weights_history).T, labels=assets)
    plt.gca().xaxis.set_major_locator(mdates.AutoDateLocator())
    plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
    plt.xlabel('Дата')
    plt.ylabel('Доля в портфеле')
    plt.title('Изменение структуры портфеля')
    plt.legend(loc='upper left')
    plt.grid(True)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.savefig(output_path)
    plt.show()

def forecast_portfolio(initial_investment, annual_contribution, years, returns_data, weights, start_date, model="historical"):
    portfolio_value = initial_investment
    portfolio_values = [portfolio_value]
    dates = [start_date]
    yearly_results = []

    if model == "historical":
        annual_returns = (returns_data + 1).resample('Y').prod().sub(1).dropna()
        for year in range(years):
            annual_return = np.sum(annual_returns.sample(n=1).iloc[0] * weights)
            portfolio_value += annual_contribution
            portfolio_value *= (1 + annual_return)
            portfolio_values.append(portfolio_value)
            dates.append(start_date + timedelta(days=365*(year+1)))
            yearly_results.append({"Год": start_date.year + year + 1, "Доходность": annual_return, "Стоимость портфеля": portfolio_value})

    return portfolio_values, dates, yearly_results

def rebalance_portfolio(initial_investment, annual_contribution, years, returns_data, initial_weights, CUSTOM_WEIGHTS=None, risk_aversion_decay=RISK_AVERSION_DECAY, start_date=None, model="historical", EXPERT_FORECAST_RETURNS=None, vol_rolling_years=VOL_ROLLING_YEARS, debug=DEBUG_REBALANCE):
    tickers = returns_data.columns.tolist()
    n = len(tickers)

    if CUSTOM_WEIGHTS:
        custom_weights_array = np.array([CUSTOM_WEIGHTS.get(a, 0.0) for a in tickers], dtype=float)
        if custom_weights_array.sum() > 0:
            custom_weights_array /= custom_weights_array.sum()
        else:
            custom_weights_array = np.array(initial_weights, dtype=float)
    else:
        custom_weights_array = np.array(initial_weights, dtype=float)

    current_weights = custom_weights_array.copy()
    if current_weights.sum() == 0:
        current_weights = np.ones(n) / n
    else:
        current_weights = current_weights / current_weights.sum()

    portfolio_value = initial_investment
    dates = [start_date]
    portfolio_values = [portfolio_value]
    weights_history = [current_weights.copy()]
    yearly_results = []

    annual_returns = (returns_data + 1).resample('Y').prod().sub(1).dropna()

    window_days = int(vol_rolling_years * TRADING_DAYS)

    for year in range(years):
        if model == "historical":
            chosen = annual_returns.sample(n=1, replace=True).iloc[0].values
            annual_return = float(np.dot(chosen, current_weights))
        elif model == "expert" and EXPERT_FORECAST_RETURNS is not None:
            expert_vec = np.array([EXPERT_FORECAST_RETURNS.get(t, 0.0) for t in tickers])
            annual_return = float(np.dot(expert_vec, current_weights))
        else:
            annual_return = 0.0

        portfolio_value = (portfolio_value + annual_contribution) * (1 + annual_return)
        dates.append(start_date + timedelta(days=365*(year+1)))
        portfolio_values.append(portfolio_value)
        yearly_results.append({"Год": start_date.year + year + 1, "Доходность": annual_return, "Стоимость портфеля": portfolio_value, "Веса": current_weights.copy()})

        if len(returns_data) >= window_days:
            vol_series = returns_data.iloc[-window_days:].std() * np.sqrt(TRADING_DAYS)
        else:
            vol_series = returns_data.std() * np.sqrt(TRADING_DAYS)
        risk_ranking = np.argsort(vol_series.values)  

        riskiest_idx = int(risk_ranking[-1])
        weight_reduction = current_weights[riskiest_idx] * (1 - risk_aversion_decay)

        current_weights[riskiest_idx] -= weight_reduction
        k = min(3, n)
        least_idxs = risk_ranking[:k]
        for idx in least_idxs:
            current_weights[idx] += weight_reduction / k

        current_weights = np.maximum(current_weights, 0.0)
        current_weights /= current_weights.sum()

        weights_history.append(current_weights.copy())

        if debug:
            print(f"Year {year+1}: riskiest={tickers[riskiest_idx]}, reduction={weight_reduction:.6f}")
            print("weights sample:", np.round(current_weights, 4))

    return portfolio_values, dates, weights_history, yearly_results

if __name__ == "__main__":
    os.makedirs(OUTPUT_FOLDER, exist_ok=True)

    df = load_data(DATA_FILE)
    if df is None:
        exit()

    selected_data = select_indices(df, SELECTED_INDICES_NAMES)
    if selected_data is None or selected_data.empty:
        print("Не выбраны индексы для анализа. Завершение работы.")
        exit()

    returns_data = select_date_range(selected_data, START_DATE_STR, END_DATE_STR)
    if returns_data is None or returns_data.empty:
        print("Не выбран диапазон дат. Завершение работы.")
        exit()

    returns = calculate_returns(returns_data)

    bounds = (MIN_ALLOCATION, MAX_ALLOCATION)

    markowitz_min_vol, markowitz_max_sharpe = markowitz_optimization(returns, risk_free_rate, bounds=bounds)
    monte_carlo_results = monte_carlo_simulation(returns, risk_free_rate, num_simulations=NUM_SIMULATIONS, bounds=bounds)

    mean_weights_mc = pd.DataFrame(monte_carlo_results[3:,:], index=returns.columns).mean(axis=1)
    mean_weights_mc /= mean_weights_mc.sum()
    mean_weights_dict = {'weights': mean_weights_mc.values}

    df_portfolios  = pd.DataFrame(monte_carlo_results.T, columns=['Volatility', 'Return', 'Sharpe'] + list(returns.columns))
    min_vol_port_mc = df_portfolios.iloc[df_portfolios['Volatility'].idxmin()]
    max_sharpe_port_mc = df_portfolios.iloc[df_portfolios['Sharpe'].idxmax()]

    plot_path = os.path.join(OUTPUT_FOLDER, "efficient_frontier.png")
    plot_efficient_frontier_and_portfolios(monte_carlo_results, mean_weights_dict, markowitz_min_vol, markowitz_max_sharpe, returns, plot_path)

    forecast_start_date = datetime.strptime(FORECAST_START_DATE_STR, '%Y-%m-%d').date()

    optimal_portfolios = {
        "Min. Risk (Markowitz)": markowitz_min_vol['x'],
        "Max. Sharpe (Markowitz)": markowitz_max_sharpe['x'],
        "Min. Risk (MC)": min_vol_port_mc.iloc[3:].values,
        "Max. Sharpe (MC)": max_sharpe_port_mc.iloc[3:].values,
        "Average Weights (MC)": mean_weights_dict['weights']
    }

    forecast_results = {}
    for name, weights in optimal_portfolios.items():
        forecast_values, forecast_dates, yearly_results = forecast_portfolio(INITIAL_DEPOSIT, ANNUAL_CONTRIBUTION, FORECAST_YEARS, returns, weights, forecast_start_date)
        forecast_results[name] = {"dates": forecast_dates, "forecast": forecast_values, "weights": weights, "yearly_results": yearly_results}

        plot_portfolio_forecast(forecast_values, forecast_dates, f"Portfolio Forecast: {name}", os.path.join(OUTPUT_FOLDER, f"forecast_{name.replace(' ', '_')}.png"))

        portfolio_values_rb, dates_rb, weights_history_rb, yearly_results_rb = rebalance_portfolio(INITIAL_DEPOSIT, ANNUAL_CONTRIBUTION, FORECAST_YEARS, returns, weights, CUSTOM_WEIGHTS=None, risk_aversion_decay=RISK_AVERSION_DECAY, start_date=forecast_start_date, model="historical", EXPERT_FORECAST_RETURNS=None, vol_rolling_years=VOL_ROLLING_YEARS, debug=DEBUG_REBALANCE)

        plot_portfolio_forecast(portfolio_values_rb, dates_rb, f"Portfolio Forecast (Rebalanced): {name}", os.path.join(OUTPUT_FOLDER, f"forecast_rebalanced_{name.replace(' ', '_')}.png"))
        plot_portfolio_structure(weights_history_rb, dates_rb, returns.columns, os.path.join(OUTPUT_FOLDER, f"portfolio_structure_rebalanced_{name.replace(' ', '_')}.png"))

        max_drawdown = calculate_max_drawdown(portfolio_values_rb)
        print(f"Max Drawdown (Rebalanced - {name}): {max_drawdown:.2%}")

        print(f"\n{name} (static):")
        df_yearly = pd.DataFrame(yearly_results)
        print(df_yearly.to_string(index=False))

        print(f"\n{name} (rebalanced):")
        df_yearly_rb = pd.DataFrame(yearly_results_rb)
        print(df_yearly_rb.to_string(index=False))

    print(f"\nResults saved to: {OUTPUT_FOLDER}")
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import scipy.optimize as sco
from datetime import datetime, timedelta
from time import time

DATA_FILE = "stocks_python.xlsx"
SHEET_NAME = "DATA"
OUTPUT_FOLDER = "Portfolio_Optimization"
SELECTED_TICKERS = ['SBER', 'FLOT', 'IRAO', 'NVTK', 'PHOR', 'MGNT', 'CHMF']

DIVIDEND_YIELDS = {
    'SBER': 0.0801,
    'FLOT': 0.0587,
    'IRAO': 0.0741,
    'NVTK': 0.0646,
    'PHOR': 0.1141,
    'MGNT': 0.0484,
    'CHMF': 0.0734,
}

RISK_FREE_RATE = 0.1361  
NUM_MC = 35_000         
MIN_ALLOCATION = 0.05
MAX_ALLOCATION = 0.25

FORECAST_YEARS = 20
INITIAL_DEPOSIT = 144_637_500
ANNUAL_CONTRIBUTION = 43_018_185

ENSEMBLE_RUNS = 180    
SEED = 42
TRADING_DAYS = 252

RISK_AVERSION_DECAY = 0.9
VOL_ROLLING_YEARS = 3

plt.rcParams["figure.figsize"] = (12, 6)

def ensure_output_folder(path):
    os.makedirs(path, exist_ok=True)

def load_data(file_path, sheet_name="DATA"):
    try:
        df = pd.read_excel(file_path, sheet_name=sheet_name, index_col=0, parse_dates=True)
        return df
    except Exception as e:
        raise RuntimeError(f"Error loading data: {e}")

def select_tickers(df, tickers):
    available = df.columns.tolist()
    selected = [t for t in tickers if t in available]
    if not selected:
        raise RuntimeError(f"No requested tickers found in data. Available: {available}")
    return df[selected]

def compute_returns_with_dividends(prices_df, dividend_yields):
    price_returns = prices_df.pct_change()
    div_daily = {}
    for t in prices_df.columns:
        if t in dividend_yields:
            ann = dividend_yields[t]
            div_daily[t] = (1 + ann) ** (1 / TRADING_DAYS) - 1
        else:
            div_daily[t] = 0.0
    div_vec = np.array([div_daily[t] for t in prices_df.columns], dtype=float)
    dividend_df = pd.DataFrame(np.tile(div_vec, (len(price_returns), 1)),
                               index=price_returns.index, columns=price_returns.columns)
    total_returns = price_returns + dividend_df
    return total_returns.dropna()

def portfolio_annualised_performance(weights, mean_returns, cov_matrix):
    ann_ret = float(np.dot(weights, mean_returns) * TRADING_DAYS)
    ann_vol = float(np.sqrt(np.dot(weights.T, np.dot(cov_matrix, weights))) * np.sqrt(TRADING_DAYS))
    return ann_vol, ann_ret

def generate_random_weights(n, min_w, max_w, rng, max_tries=5000):
    if min_w * n > 1.0 + 1e-12:
        raise ValueError("min_w * n > 1: impossible constraints")
    for _ in range(max_tries):
        base = np.full(n, min_w)
        remaining = 1.0 - base.sum()
        if remaining <= 0:
            w = base.copy()
        else:
            y = rng.gamma(1.0, 1.0, size=n)
            y = y / y.sum()
            w = base + remaining * y
        if np.all(w <= max_w + 1e-12):
            return w
    raise RuntimeError("Failed to generate random weights respecting bounds")

def monte_carlo_once(returns_df, rf, num_simulations=100_000, min_w=0.05, max_w=0.25, seed=0):
    rng = np.random.default_rng(seed)
    mean_returns = returns_df.mean()
    cov = returns_df.cov()
    num_assets = len(returns_df.columns)
    results = np.zeros((3 + num_assets, num_simulations))
    for i in range(num_simulations):
        w = generate_random_weights(num_assets, min_w, max_w, rng)
        p_vol = np.sqrt(np.dot(w.T, np.dot(cov.values, w))) * np.sqrt(TRADING_DAYS)
        p_ret = float(np.dot(w, mean_returns.values) * TRADING_DAYS)
        results[0, i] = p_vol
        results[1, i] = p_ret
        results[2, i] = (p_ret - rf) / p_vol if p_vol > 0 else -1e9
        results[3:, i] = w
    return results

def markowitz_min_variance(returns_df, bounds=(0.05, 0.25)):
    mean_returns = returns_df.mean()
    cov = returns_df.cov()
    n = len(returns_df.columns)
    args = (mean_returns, cov)
    constraints = ({'type': 'eq', 'fun': lambda x: np.sum(x) - 1})
    bnds = tuple((bounds[0], bounds[1]) for _ in range(n))
    def portfolio_vol(weights, mean_returns, cov_matrix):
        return portfolio_annualised_performance(weights, mean_returns, cov_matrix)[0]
    x0 = np.array([1.0 / n] * n)
    opt = sco.minimize(portfolio_vol, x0, args=args, method='SLSQP', bounds=bnds, constraints=constraints, options={'maxiter':1000})
    if not opt.success:
        raise RuntimeError("Markowitz minimization failed: " + str(opt.message))
    w = np.maximum(opt.x, 0.0)
    w = w / w.sum()
    return w

def compute_max_drawdown(values):
    arr = np.array(values, dtype=float)
    cummax = np.maximum.accumulate(arr)
    drawdowns = (cummax - arr) / cummax
    return float(np.nanmax(drawdowns))

def forecast_with_rebalance(initial_investment, annual_contribution, years, returns_df, initial_weights,
                            start_date, seed=0, risk_aversion_decay=RISK_AVERSION_DECAY, vol_rolling_years=VOL_ROLLING_YEARS):
    rng = np.random.default_rng(seed)
    tickers = returns_df.columns.tolist()
    n = len(tickers)
    w = np.array(initial_weights, dtype=float)
    if np.sum(w) == 0:
        w = np.ones(n) / n
    else:
        w = np.maximum(w, 0.0)
        w = w / w.sum()

    portfolio_value = float(initial_investment)
    portfolio_values = [portfolio_value]
    dates = [start_date]
    weights_history = [w.copy()]
    yearly_results = []

    annual_returns_table = (returns_df + 1).resample('YE').prod().sub(1).dropna()
    window_days = int(vol_rolling_years * TRADING_DAYS)

    for year in range(years):
        if len(annual_returns_table) == 0:
            sampled = returns_df.mean().values * TRADING_DAYS
        else:
            sampled = annual_returns_table.sample(n=1, random_state=seed + year).iloc[0].values

        portfolio_value = (portfolio_value + annual_contribution) * (1.0 + float(np.dot(sampled, w)))
        new_date = start_date + timedelta(days=365 * (year + 1))
        dates.append(new_date)
        portfolio_values.append(portfolio_value)
        yearly_results.append({"Year": start_date.year + year + 1, "Return": float(np.dot(sampled, w)), "PortfolioValue": portfolio_value, "Weights": w.copy()})

        if len(returns_df) >= window_days:
            vol_series = returns_df.iloc[-window_days:].std() * np.sqrt(TRADING_DAYS)
        else:
            vol_series = returns_df.std() * np.sqrt(TRADING_DAYS)
        risk_ranking = np.argsort(vol_series.values)
        riskiest_idx = int(risk_ranking[-1])
        weight_reduction = w[riskiest_idx] * (1 - risk_aversion_decay)
        w[riskiest_idx] -= weight_reduction
        k = min(3, n)
        least_idxs = risk_ranking[:k]
        for idx in least_idxs:
            w[idx] += weight_reduction / k
        w = np.maximum(w, 0.0)
        w = w / w.sum()
        weights_history.append(w.copy())

    return portfolio_values, dates, weights_history, yearly_results

def ensemble_max_sharpe_forecasts(ensemble_runs, num_mc_per_run, returns_df, rf, initial_deposit, annual_contrib, years, forecast_start_date, seed_base=SEED):
    n_years = years
    trajs = np.zeros((ensemble_runs, n_years + 1), dtype=float)
    weight_vectors = []
    weights_histories = []
    runtimes = []
    for r in range(ensemble_runs):
        t0 = time()
        seed_run = seed_base + r
        mc = monte_carlo_once(returns_df, rf, num_simulations=num_mc_per_run, min_w=MIN_ALLOCATION, max_w=MAX_ALLOCATION, seed=seed_run)
        df_mc = pd.DataFrame(mc.T, columns=['Vol','Return','Sharpe'] + list(returns_df.columns))
        idx_best = int(df_mc['Sharpe'].idxmax())
        best_weights = df_mc.loc[idx_best, list(returns_df.columns)].values.astype(float)
        if np.sum(best_weights) <= 0:
            best_weights = np.ones(len(best_weights)) / len(best_weights)
        else:
            best_weights = np.maximum(best_weights, 0.0)
            best_weights = best_weights / best_weights.sum()
        pv, dates, weights_hist, yearly_results = forecast_with_rebalance(initial_deposit, annual_contrib, years, returns_df, best_weights, forecast_start_date, seed=seed_run)
        trajs[r, :] = np.array(pv)
        weight_vectors.append(best_weights)
        weights_histories.append(weights_hist)
        runtimes.append(time() - t0)
        if (r + 1) % 10 == 0 or r == 0 or r == ensemble_runs - 1:
            avg_t = np.mean(runtimes) if runtimes else 0.0
            rem = (ensemble_runs - (r + 1)) * avg_t
            print(f"Ensemble run {r+1}/{ensemble_runs} done (run_time={runtimes[-1]:.1f}s). Est. remaining {rem/60:.1f} min")
    return trajs, weight_vectors, weights_histories

def main():
    ensure_output_folder(OUTPUT_FOLDER)

    df_all = load_data(DATA_FILE, sheet_name=SHEET_NAME)
    selected = select_tickers(df_all, SELECTED_TICKERS)
    returns = compute_returns_with_dividends(selected, DIVIDEND_YIELDS)

    last = selected.index.max()
    forecast_start_date = (last + pd.Timedelta(days=1)).date()

    print("Computing Markowitz min-variance portfolio...")
    markowitz_weights = markowitz_min_variance(returns, bounds=(MIN_ALLOCATION, MAX_ALLOCATION))
    print("Markowitz min-variance weights (in %):")
    for t, w in zip(returns.columns, markowitz_weights):
        print(f"  {t}: {w*100:.2f}%")

    print("Doing one Monte Carlo (single) to show example (this is heavy)...")
    mc_single = monte_carlo_once(returns, RISK_FREE_RATE, num_simulations=NUM_MC, min_w=MIN_ALLOCATION, max_w=MAX_ALLOCATION, seed=SEED)
    df_mc_single = pd.DataFrame(mc_single.T, columns=['Vol','Return','Sharpe'] + list(returns.columns))
    idx_max_sh = int(df_mc_single['Sharpe'].idxmax())
    max_sh_weights_single = df_mc_single.loc[idx_max_sh, list(returns.columns)].values
    max_sh_weights_single = np.maximum(max_sh_weights_single, 0.0)
    max_sh_weights_single = max_sh_weights_single / max_sh_weights_single.sum()

    print("Example MaxSharpe weights (single run) in %:")
    for t, w in zip(returns.columns, max_sh_weights_single):
        print(f"  {t}: {w*100:.2f}%")

    print(f"\nStarting ensemble of {ENSEMBLE_RUNS} runs, each with {NUM_MC} MC sims. This may take long.")
    t_ens0 = time()
    trajs, weight_vectors, weights_histories = ensemble_max_sharpe_forecasts(ENSEMBLE_RUNS, NUM_MC, returns, RISK_FREE_RATE, INITIAL_DEPOSIT, ANNUAL_CONTRIBUTION, FORECAST_YEARS, forecast_start_date, seed_base=SEED)
    t_ens1 = time()
    print(f"Ensemble completed in {(t_ens1 - t_ens0)/60:.2f} minutes")

    perciles = [5, 25, 50, 75, 95]
    perc_data = {}
    years_labels = [forecast_start_date.year + i for i in range(FORECAST_YEARS + 1)]
    for p in perciles:
        perc_data[f"p{p}"] = np.percentile(trajs, p, axis=0)
    df_perc = pd.DataFrame({k: v for k, v in perc_data.items()}, index=years_labels)
    df_perc.index.name = "Year"
    df_perc.to_csv(os.path.join(OUTPUT_FOLDER, "ensemble_percentiles_by_year.csv"))

    df_trajs = pd.DataFrame(trajs, columns=[str(y) for y in years_labels])
    df_trajs.to_csv(os.path.join(OUTPUT_FOLDER, "ensemble_trajectories.csv"), index=False)

    years_dt = [forecast_start_date + timedelta(days=365 * i) for i in range(FORECAST_YEARS + 1)]
    median = df_perc['p50'].values
    p5 = df_perc['p5'].values
    p95 = df_perc['p95'].values
    p25 = df_perc['p25'].values
    p75 = df_perc['p75'].values

    plt.figure(figsize=(12, 7))
    plt.plot(years_dt, median, label='Медиана (50%)', linewidth=2, color='black')
    plt.fill_between(years_dt, p5, p95, color='lightgray', alpha=0.6, label='5-95% ДИ')
    plt.fill_between(years_dt, p25, p75, color='silver', alpha=0.8, label='25-75% ДИ')
    plt.scatter(years_dt, trajs.mean(axis=0), s=12, alpha=0.6, label='Среднее', color='tab:blue')
    plt.xlabel('Год')
    plt.ylabel('Стоимость портфеля, в руб.')
    plt.title(f'Прогноз стоимости портфеля (Max-Sharpe MC) — {ENSEMBLE_RUNS} runs x {NUM_MC} sims')
    plt.legend()
    plt.grid(True)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.savefig(os.path.join(OUTPUT_FOLDER, f'ensemble_forecast_CI_{ENSEMBLE_RUNS}x{NUM_MC}.png'))
    plt.show()

    final_vals = trajs[:, -1]
    median_final = df_perc['p50'].values[-1]
    idx_closest = int(np.argmin(np.abs(final_vals - median_final)))
    print(f"\nIndex of ensemble member closest to median final value: {idx_closest}")
    chosen_best_weights = weight_vectors[idx_closest]
    chosen_weights_history = weights_histories[idx_closest]  # list length years+1
    initial_weights_chosen = chosen_best_weights
    final_weights_chosen = chosen_weights_history[-1]

    print("\nChosen ensemble member weights (initial, %):")
    for t, w in zip(returns.columns, initial_weights_chosen):
        print(f"  {t}: {w*100:.2f}%")

    print("\nChosen ensemble member weights (final after rebalance years, %):")
    for t, w in zip(returns.columns, final_weights_chosen):
        print(f"  {t}: {w*100:.2f}%")

    weights_array = np.array(chosen_weights_history) 
    dates_stack = [forecast_start_date + timedelta(days=365 * i) for i in range(len(weights_array))]
    plt.figure(figsize=(12, 6))
    plt.stackplot(dates_stack, weights_array.T, labels=returns.columns)
    plt.gca().xaxis.set_major_locator(mdates.AutoDateLocator())
    plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
    plt.title('Weights history for ensemble member closest to median final value')
    plt.xlabel('Date')
    plt.ylabel('Weight')
    plt.legend(loc='upper left', bbox_to_anchor=(1.05, 1))
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.savefig(os.path.join(OUTPUT_FOLDER, f'weights_history_median_member_idx{idx_closest}.png'))
    plt.show()

    df_weights_hist = pd.DataFrame(weights_array, columns=returns.columns, index=[d.strftime('%Y-%m-%d') for d in dates_stack])
    df_weights_hist.to_csv(os.path.join(OUTPUT_FOLDER, f'weights_history_median_member_idx{idx_closest}.csv'))

    final_row = df_perc.iloc[-1]
    print(f"\nFinal year ({df_perc.index[-1]}) percentiles (rubles):")
    for p in [5,25,50,75,95]:
        print(f"  p{p}: {final_row[f'p{p}']:.2f}")

    summary = {
        "ENSEMBLE_RUNS": ENSEMBLE_RUNS,
        "NUM_MC_per_run": NUM_MC,
        "final_year": df_perc.index[-1],
        "p5_final": float(final_row['p5']),
        "p25_final": float(final_row['p25']),
        "p50_final": float(final_row['p50']),
        "p75_final": float(final_row['p75']),
        "p95_final": float(final_row['p95']),
        "median_member_index": idx_closest
    }
    pd.DataFrame([summary]).to_csv(os.path.join(OUTPUT_FOLDER, "ensemble_summary_with_median_member.csv"), index=False)
    print(f"\nAll outputs saved to folder: {OUTPUT_FOLDER}")

if __name__ == "__main__":
    main()
```

```python
import time
from dataclasses import dataclass
from typing import List

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import requests
from scipy.optimize import minimize

# ===========================
# 1. НАСТРОЙКИ
# ===========================

FROM_DATE = "2020-01-01"           # от какой даты брать историю цен
RISK_FREE_RATE = 0.1361            # 13.61% в ДОЛЯХ для Шарпа!
ANNUAL_INVESTMENT = -4_772_200     # сколько в год вкладываем в портфель
INITIAL_INVESTMENT = 60_487_500    # ИСХОДНАЯ сумма в облигациях
YEARS = 20                         # горизонт инвестирования
YIELD_FLOOR = 0.08                 # минимальная ожидаемая доходность облигаций (8% годовых в долях)
N_REBAL_PATHS = 10_000             # число сценариев в симуляции ребалансировки

np.random.seed(42)                 # чтобы результаты Monte Carlo воспроизводились


@dataclass
class Asset:
    secid: str        # тикер на MOEX / ISIN-код в ISS
    name: str         # человекочитаемое имя инструмента
    asset_class: str  # "bonds"
    market: str       # для ISS: "bonds"


# -------------------------------
#  ЮНИВЕРС ОБЛИГАЦИЙ (8 ШТУК)
# -------------------------------

BONDS: List[Asset] = [
    # Государственные облигации (ОФЗ, де-факто суверенный AAA)
    Asset(
        secid="SU26238RMFS4",
        name="ОФЗ-ПД 26238 15.05.2041",
        asset_class="bonds",
        market="bonds",
    ),
    Asset(
        secid="SU26239RMFS2",
        name="ОФЗ-ПД 26239 23.07.2031",
        asset_class="bonds",
        market="bonds",
    ),
    Asset(
        secid="SU26240RMFS0",
        name="ОФЗ-ПД 26240 30.07.2036",
        asset_class="bonds",
        market="bonds",
    ),
    Asset(
        secid="SU26241RMFS8",
        name="ОФЗ-ПД 26241 17.11.2032",
        asset_class="bonds",
        market="bonds",
    ),

    # ===== Корпоративные облигации высокого качества (AA/AAA) =====
    Asset(
        secid="RU000A109SK6",
        name="МТС 001P-27, AAA(RU)",
        asset_class="bonds",
        market="bonds",
    ),
    Asset(
        secid="RU000A100JC1",
        name="ЕАБР 001P-04, AAA(RU)",
        asset_class="bonds",
        market="bonds",
    ),
    Asset(
        secid="RU000A101574",
        name="ЕАБР 001P-05, AAA(RU)",
        asset_class="bonds",
        market="bonds",
    ),
]


# ===========================
# 2. ЗАГРУЗКА ИСТОРИИ С MOEX
# ===========================

def fetch_history(asset: Asset, from_date: str, max_retries: int = 5) -> pd.Series:
    """
    История цен по облигации с MOEX ISS.
    Индекс — дата (уникальная), значения — CLOSE.
    При сетевых ошибках делает несколько попыток; если не вышло — пустая Series.
    """
    base_url = (
        f"https://iss.moex.com/iss/history/engines/stock/"
        f"markets/{asset.market}/securities/{asset.secid}.json"
    )

    params = {
        "from": from_date,
        "iss.meta": "off",
        "history.columns": "TRADEDATE,SECID,CLOSE",
    }

    all_rows = []
    start = 0

    while True:
        params["start"] = start

        last_error = None
        for attempt in range(max_retries):
            try:
                resp = requests.get(base_url, params=params, timeout=30)
                resp.raise_for_status()
                data = resp.json()
                break
            except (requests.exceptions.ReadTimeout,
                    requests.exceptions.ConnectionError) as e:
                last_error = e
                wait_sec = 2 * (attempt + 1)
                print(
                    f"Сетевая ошибка по {asset.secid} "
                    f"(попытка {attempt + 1}/{max_retries}): {e}. "
                    f"Ждём {wait_sec} сек..."
                )
                time.sleep(wait_sec)
        else:
            print(f"Не удалось загрузить историю по {asset.secid} ({asset.name}). "
                  f"Пропускаем. Последняя ошибка: {last_error}")
            return pd.Series(dtype=float)

        rows = data.get("history", {}).get("data", [])
        if not rows:
            break

        all_rows.extend(rows)
        start += len(rows)

    if not all_rows:
        print(f"Нет истории по инструменту {asset.secid} ({asset.name})")
        return pd.Series(dtype=float)

    cols = data["history"]["columns"]
    df = pd.DataFrame(all_rows, columns=cols)
    df = df[["TRADEDATE", "CLOSE"]].dropna()
    df["TRADEDATE"] = pd.to_datetime(df["TRADEDATE"])

    # одна цена на день (берём последнюю)
    df = df.groupby("TRADEDATE", as_index=False)["CLOSE"].last()
    df.set_index("TRADEDATE", inplace=True)
    df = df[~df.index.duplicated(keep="last")]
    df.sort_index(inplace=True)

    return df["CLOSE"].astype(float)


def build_price_dataframe(assets: List[Asset], from_date: str) -> pd.DataFrame:
    """
    Собираем цены по всем облигациям в один DataFrame.
    Не выкидываем строки с NaN — у разных бумаг разные периоды торгов.
    """
    price_dict = {}
    for a in assets:
        print(f"Загружаю историю по {a.secid} ({a.name})...")
        s = fetch_history(a, from_date=from_date)
        if s.empty:
            print(f"   * пропускаю {a.secid}: нет данных")
            continue
        price_dict[a.secid] = s

    if len(price_dict) < 2:
        raise ValueError("Нужно минимум две облигации с историей для оптимизации портфеля.")

    prices = pd.concat(price_dict, axis=1).sort_index()
    return prices


def annualized_return_and_cov(prices: pd.DataFrame, min_obs: int = 60):
    """
    Считаем дневные доходности, годовые средние и ковариации.
    Не выбрасываем строки с NaN, фильтруем бумаги с короткой историей.
    """
    daily_ret = prices.pct_change()

    # оставляем только те бумаги, у которых достаточно наблюдений
    valid_cols = [c for c in daily_ret.columns if daily_ret[c].count() >= min_obs]
    daily_ret = daily_ret[valid_cols]

    if len(valid_cols) < 2:
        raise ValueError("После фильтрации по длине истории осталось меньше двух облигаций.")

    mean_daily = daily_ret.mean(skipna=True)
    cov_daily = daily_ret.cov(min_periods=min_obs)

    mean_annual = mean_daily * 252
    cov_annual = cov_daily * 252
    return daily_ret, mean_annual, cov_annual


# ===========================
# 3. МАРКОВИЦ: MAX SHARPE & MIN VAR
# ===========================

def portfolio_performance(weights, mean_returns, cov_matrix, rf=0.0):
    weights = np.array(weights)
    r = float(np.dot(weights, mean_returns))
    vol = float(np.sqrt(weights.T @ cov_matrix @ weights))
    sharpe = (r - rf) / vol if vol > 0 else 0.0
    return r, vol, sharpe


def max_sharpe_ratio(mean_returns: np.ndarray, cov_matrix: np.ndarray, rf=0.0):
    n = len(mean_returns)

    def neg_sharpe(w):
        _, _, s = portfolio_performance(w, mean_returns, cov_matrix, rf)
        return -s

    constraints = ({"type": "eq", "fun": lambda w: np.sum(w) - 1},)
    bounds = tuple((0.0, 1.0) for _ in range(n))
    x0 = np.ones(n) / n

    res = minimize(neg_sharpe, x0, method="SLSQP", bounds=bounds, constraints=constraints)
    if not res.success:
        raise RuntimeError("Оптимизация max Sharpe не сошлась: " + res.message)

    w = res.x
    r, vol, s = portfolio_performance(w, mean_returns, cov_matrix, rf)
    return w, r, vol, s


def min_variance_portfolio(mean_returns: np.ndarray, cov_matrix: np.ndarray):
    n = len(mean_returns)

    def port_vol(w):
        _, vol, _ = portfolio_performance(w, mean_returns, cov_matrix)
        return vol

    constraints = ({"type": "eq", "fun": lambda w: np.sum(w) - 1},)
    bounds = tuple((0.0, 1.0) for _ in range(n))
    x0 = np.ones(n) / n

    res = minimize(port_vol, x0, method="SLSQP", bounds=bounds, constraints=constraints)
    if not res.success:
        raise RuntimeError("Оптимизация min-var не сошлась: " + res.message)

    w = res.x
    r, vol, s = portfolio_performance(w, mean_returns, cov_matrix)
    return w, r, vol, s


# ===========================
# 4. МОНТЕ-КАРЛО И ГРАФИК
# ===========================

def monte_carlo_portfolios(mean_returns, cov_matrix,
                           n_portfolios=40_000, risk_free_rate=0.0):
    """
    Генерируем случайные портфели, считаем доходность, риск и Sharpe.
    Возвращает dict с ключами: 'returns', 'vols', 'sharpes'.
    """
    n = len(mean_returns)
    rets = np.zeros(n_portfolios)
    vols = np.zeros(n_portfolios)
    sharpes = np.zeros(n_portfolios)

    for i in range(n_portfolios):
        w = np.random.random(n)
        w /= np.sum(w)
        r, v, s = portfolio_performance(w, mean_returns, cov_matrix, risk_free_rate)
        rets[i], vols[i], sharpes[i] = r, v, s

    return {"returns": rets, "vols": vols, "sharpes": sharpes}


def plot_monte_carlo_cloud(mc,
                           r_eq, vol_eq,
                           r_mv, vol_mv,
                           r_ms, vol_ms):
    """
    Рисует:
      - облако случайных портфелей (Монте-Карло)
      - красный крест: равные доли
      - зелёная звезда: мин. риск
      - синий ромб: max Sharpe
    """
    plt.figure(figsize=(10, 7))

    # облако
    plt.scatter(mc["vols"] * 100, mc["returns"] * 100,
                c=mc["sharpes"], s=4, alpha=0.4, cmap="viridis")

    # равные доли
    plt.scatter(vol_eq * 100, r_eq * 100,
                marker="X", s=200, color="red", label="Средние доли")

    # мин. риск
    plt.scatter(vol_mv * 100, r_mv * 100,
                marker="*", s=200, color="green", label="Мин. риск")

    # макс. Sharpe
    plt.scatter(vol_ms * 100, r_ms * 100,
                marker="D", s=200, color="blue", edgecolors="black", label="Портфель max Sharpe")

    plt.title("Облако портфелей (Монте-Карло, облигации)")
    plt.xlabel("Годовая волатильность, %")
    plt.ylabel("Ожидаемая доходность, %")
    plt.grid(True, alpha=0.3)
    plt.legend(loc="best")
    cbar = plt.colorbar()
    cbar.set_label("Sharpe ratio")
    plt.tight_layout()
    plt.show()


def plot_initial_portfolio_diversification(weights, tickers, bonds):
    """
    Круговая диаграмма по долям бумаг в портфеле.
    Нулевые веса не рисуем, чтобы не было наложения подписей.
    """
    labels = []
    sizes = []
    for secid, w in zip(tickers, weights):
        if w <= 1e-6:
            # пропускаем нулевые веса, чтобы не плодить тонкие клинья с подписями
            continue
        asset = next(a for a in bonds if a.secid == secid)
        labels.append(asset.name)
        sizes.append(w)

    plt.figure(figsize=(8, 8))
    plt.pie(sizes, labels=labels, autopct="%1.1f%%", startangle=90)
    plt.title("Диверсификация облигационного портфеля")
    plt.axis("equal")  # круг, а не овал
    plt.tight_layout()
    plt.show()



# ===========================
# 5. СИМУЛЯЦИЯ С РЕБАЛАНСИРОВКОЙ
# ===========================

def simulate_rebalancing_paths(mean_returns, cov_matrix, target_weights,
                               annual_investment, years, n_paths=10_000,
                               initial_investment=0.0):
    """
    Monte Carlo: ежегодная ребалансировка к target_weights.
    Каждый год:
      - имеем стартовый капитал initial_investment (в начале 0-го года);
      - добавляем annual_investment;
      - ребалансируем к target_weights;
      - генерируем случайный вектор доходностей ~ N(mean, cov);
      - применяем доходности к портфелю.
    Возвращает массив конечных капиталов.
    """
    mean = np.array(mean_returns)
    cov = np.array(cov_matrix)
    n_assets = len(mean)

    L = np.linalg.cholesky(cov)  # для генерации коррелированных доходностей

    final_capitals = np.zeros(n_paths)

    for path in range(n_paths):
        total_capital = float(initial_investment)
        holdings = np.zeros(n_assets)

        # начальное распределение стартового капитала
        if total_capital > 0:
            holdings = total_capital * target_weights

        for year in range(years):
            # 1. ежегодный взнос + ребалансировка
            if annual_investment > 0:
                total_capital = holdings.sum() + annual_investment
            else:
                total_capital = holdings.sum()

            holdings = total_capital * target_weights

            # 2. годовая доходность
            z = np.random.normal(size=n_assets)
            yearly_returns = mean + L @ z  # вектор годовых доходностей

            # 3. применяем доходности
            holdings = holdings * (1 + yearly_returns)
            total_capital = holdings.sum()

        final_capitals[path] = total_capital

    return final_capitals


# ===========================
# 6. АННУИТЕТ (ДЕТЕРМИНИСТИКА)
# ===========================

def future_value_annuity(payment: float, r: float, n: int) -> float:
    """
    Будущая стоимость аннуитета (серия равных платежей payment раз в год).
    """
    if r == 0:
        return payment * n
    return payment * ((1 + r) ** n - 1) / r

def plot_deterministic_capital_path(initial_investment: float,
                                    annual_investment: float,
                                    r: float,
                                    years: int,
                                    start_year: int | None = None,
                                    title_suffix: str = "облигационный портфель, без учёта инфляции"):
    """
    Рисует детерминированную траекторию капитала по годам.
    Если start_year задан, по оси X будут календарные годы (целые).
    """
    periods = list(range(1, years + 1))  # 1..N

    capitals_total = []
    capitals_from_initial = []
    capitals_from_contributions = []

    for n in periods:
        fv_init = initial_investment * (1 + r) ** n
        fv_ann = future_value_annuity(annual_investment, r, n)
        capitals_total.append(fv_init + fv_ann)
        capitals_from_initial.append(fv_init)
        capitals_from_contributions.append(fv_ann)

    if start_year is None:
        x_vals = periods
        x_label = "Год"
        x_ticks = periods
    else:
        x_vals = [start_year + p - 1 for p in periods]  # напр. 2026..2045
        x_label = "Календарный год"
        x_ticks = x_vals

    plt.figure(figsize=(10, 6))
    plt.plot(x_vals, capitals_total, marker="o", label="Совокупный капитал")
    plt.plot(x_vals, capitals_from_initial, linestyle="--", label="Только стартовая сумма")
    plt.plot(x_vals, capitals_from_contributions, linestyle=":", label="Только ежегодные взносы")

    plt.xlabel(x_label)
    plt.ylabel("Размер капитала, ₽")
    plt.title(f"Детерминированный рост капитала ({title_suffix})")
    plt.grid(True, alpha=0.3)
    plt.legend()

    # 🔹 фиксируем ПОДПИСИ по оси X только в целых годах
    plt.xticks(x_ticks, [str(int(x)) for x in x_ticks])

    plt.tight_layout()
    plt.show()



# ===========================
# 7. MAIN
# ===========================

def main():
    bonds = BONDS

    # 1) История цен и статистика
    prices = build_price_dataframe(bonds, FROM_DATE)
    daily_ret, mean_raw, cov_annual = annualized_return_and_cov(prices, min_obs=60)

    # проверим, что есть хоть одна нормальная доходность
    if mean_raw.dropna().empty:
        raise ValueError("По выбранным облигациям нет нормальной истории доходностей. "
                         "Попробуй другой FROM_DATE или другой набор бумаг.")

    # сдвигаем доходности вверх, чтобы минимум был не ниже YIELD_FLOOR
    shift_base = mean_raw.dropna().min()
    shift = max(YIELD_FLOOR - shift_base, 0)
    mean_annual = mean_raw + shift

    # выкидываем бумаги, у которых доходность всё равно NaN
    nan_assets = mean_annual[mean_annual.isna()].index
    if len(nan_assets) > 0:
        print("\nЭти облигации исключены из оптимизации (слишком мало/кривая история):")
        for secid in nan_assets:
            asset = next(a for a in bonds if a.secid == secid)
            print(f"   - {asset.name} ({secid})")

        mean_annual = mean_annual.dropna()
        cov_annual = cov_annual.loc[mean_annual.index, mean_annual.index]

    if len(mean_annual) < 2:
        raise ValueError("После фильтрации осталось меньше двух облигаций — "
                         "оптимизировать портфель нельзя.")

    tickers = list(mean_annual.index)
    print("\nГодовые ожидаемые доходности по облигациям (с учётом пола):")
    for secid, mu in mean_annual.items():
        asset = next(a for a in bonds if a.secid == secid)
        print(f"- {asset.name:35s} ({secid}): {mu*100:6.2f}%")

    # 2) Портфель равных долей (для точки "Средние доли")
    n = len(mean_annual)
    w_eq = np.ones(n) / n
    r_eq, vol_eq, sharpe_eq = portfolio_performance(
        w_eq, mean_annual.values, cov_annual.values, RISK_FREE_RATE
    )

    # 3) Портфель максимального Sharpe
    w_ms, r_ms, vol_ms, sharpe_ms = max_sharpe_ratio(
        mean_annual.values, cov_annual.values, RISK_FREE_RATE
    )

    print("\n=== Портфель максимального Sharpe (только облигации) ===")
    print(f"Доходность:    {r_ms*100:6.2f}% годовых")
    print(f"Волатильность: {vol_ms*100:6.2f}% годовых")
    print(f"Sharpe:        {sharpe_ms:6.2f}")
    for t, w in zip(tickers, w_ms):
        asset = next(a for a in bonds if a.secid == t)
        print(f"  {asset.name:35s} ({t}): {w*100:5.2f}%")

    # График диверсификации начального портфеля (max Sharpe)
    plot_initial_portfolio_diversification(w_ms, tickers, bonds)

    # 4) Портфель минимальной дисперсии
    w_mv, r_mv, vol_mv, sharpe_mv = min_variance_portfolio(
        mean_annual.values, cov_annual.values
    )

    print("\n=== Портфель минимальной дисперсии (только облигации) ===")
    print(f"Доходность:    {r_mv*100:6.2f}% годовых")
    print(f"Волатильность: {vol_mv*100:6.2f}% годовых")
    print(f"Sharpe:        {sharpe_mv:6.2f}")
    for t, w in zip(tickers, w_mv):
        asset = next(a for a in bonds if a.secid == t)
        print(f"  {asset.name:35s} ({t}): {w*100:5.2f}%")

    # 5) Монте-Карло по облигациям
    print("\nЗапускаю Монте-Карло симуляцию случайных портфелей...")
    mc = monte_carlo_portfolios(
        mean_annual.values,
        cov_annual.values,
        n_portfolios=40_000,
        risk_free_rate=RISK_FREE_RATE,
    )

    # 6) График Монте-Карло
    plot_monte_carlo_cloud(
        mc,
        r_eq, vol_eq,
        r_mv, vol_mv,
        r_ms, vol_ms,
    )

    # 7) Детерминированный расчёт будущего капитала (аннуитет + стартовый капитал)
    fv_from_contributions = future_value_annuity(ANNUAL_INVESTMENT, r_ms, YEARS)
    fv_from_initial = INITIAL_INVESTMENT * (1 + r_ms) ** YEARS
    fv_capital_det_total = fv_from_initial + fv_from_contributions
    annual_income_det = fv_capital_det_total * r_ms

    print(f"\n=== Будущая стоимость капитала (детерминированно, облигационный портфель) ===")
    print(f"Начальная сумма инвестиций:     {INITIAL_INVESTMENT:,.0f} ₽")
    print(f"Ежегодный взнос:               {ANNUAL_INVESTMENT:,.0f} ₽")
    print(f"Ожидаемая доходность портфеля: {r_ms * 100:6.2f}% годовых")
    print(f"Горизонт:                      {YEARS} лет")
    print(f"Ожидаемый капитал (всего):     {fv_capital_det_total:,.0f} ₽")
    print(f"  из них рост стартовой суммы: {fv_from_initial:,.0f} ₽")
    print(f"  из них за счёт взносов:      {fv_from_contributions:,.0f} ₽")
    print(f"Ожидаемый пассивный доход:     {annual_income_det:,.0f} ₽ в год")

    # График детерминированного роста капитала (по оси X — годы до 2045)
    # если старт считать с 2026 года:
    plot_deterministic_capital_path(
        INITIAL_INVESTMENT,
        ANNUAL_INVESTMENT,
        r_ms,
        YEARS,
        start_year=2026,  # получишь тики 2026, 2027, ..., 2045
        title_suffix="облигационный портфель, без учёта инфляции"
    )

    # 8) Monte Carlo с ежегодной ребалансировкой к портфелю max Sharpe
    print("\nЗапускаю симуляцию с ежегодной ребалансировкой к портфелю max Sharpe...")
    final_caps = simulate_rebalancing_paths(
        mean_annual.values,
        cov_annual.values,
        target_weights=w_ms,
        annual_investment=ANNUAL_INVESTMENT,
        years=YEARS,
        n_paths=N_REBAL_PATHS,
        initial_investment=INITIAL_INVESTMENT,
    )

    mean_cap = np.mean(final_caps)
    median_cap = np.median(final_caps)
    p10 = np.percentile(final_caps, 10)
    p90 = np.percentile(final_caps, 90)

    print("\n=== Результаты Monte Carlo с ребалансировкой (портфель max Sharpe) ===")
    print(f"Сценариев:                      {N_REBAL_PATHS}")
    print(f"Начальная сумма инвестиций:     {INITIAL_INVESTMENT:,.0f} ₽")
    print(f"Ежегодный взнос:                {ANNUAL_INVESTMENT:,.0f} ₽")
    print(f"Средний конечный капитал:       {mean_cap:,.0f} ₽")
    print(f"Медианный капитал:              {median_cap:,.0f} ₽")
    print(f"10-й перцентиль (пессимистично):{p10:,.0f} ₽")
    print(f"90-й перцентиль (оптимистично): {p90:,.0f} ₽")


if __name__ == "__main__":
    main()
```
